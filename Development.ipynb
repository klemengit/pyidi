```python
import os
import sys
sys.path.insert(0, os.path.realpath('__file__'))

import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import pickle
import pyidi
matplotlib.use("TkAgg")
```

```python
filename_2 = 'data/data_synthetic.cih'
video = pyidi.VideoReader(filename_2)

if os.path.exists(r"D:\phdpy\pyIDI\data\temp_file\points.pkl"):
    print('Loading points...')
    inds_2 = pickle.load(open(r"D:\phdpy\pyIDI\data\temp_file\points.pkl", 'rb'))
else:
    print('Defining points...')
    inds_2 = np.array([
        [ 31,  35],
        [ 91,  35],
        [ 31, 215],
        [ 91, 215],
        [ 31, 126],
        [ 91, 126],
        [ 66,  71],
        [ 35,  71],
        [ 95,  71],
        [ 66, 191],
        [ 35, 191],
        [ 95, 191]
    ])


idi = pyidi.DIC(video)
idi.configure()
idi.set_points(inds_2)
```

```python
disp = idi.get_displacements()
```

```python
plt.figure()
plt.plot(disp[0, :, 1] - disp[0, 0, 1], label='calc')
plt.plot(disp[0, :, 0] - disp[0, 0, 0], label='calc')
plt.legend()
plt.show()
```
